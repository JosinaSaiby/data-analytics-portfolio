Problem Statement

KSRTC operates buses across different routes with varying distances, passenger numbers, ticket fares, and fuel costs. Analyzing this data manually can make it difficult to identify high-demand routes, estimate revenue, and understand operational costs. This project uses Python, NumPy, Pandas, Matplotlib, and Seaborn to analyze KSRTC bus data and find useful patterns and insights.

```python
# Importing required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Reading the KSRTC dataset
df = pd.read_csv("/content/Untitled spreadsheet - Sheet1.csv")
df.head()
```

```python
# Checking the shape of the dataset
df.shape
```

```python
# Checking the information of the dataset
df.info()
```

```python
# Checking for missing values

df.isnull().sum()
```

```python
# Checking for duplicate rows

df.duplicated().sum()
```

```python
# Displaying statistical summary of numerical columns

df.describe()
```

```python
# Displaying all column names

df.columns
```

```python
# Calculating the total number of passengers

df['Passengers'].sum()
```

```python
# Calculating the average number of passengers

df['Passengers'].mean()
```

```python
# Finding the route with the highest number of passengers

df.loc[df['Passengers'].idxmax()]
```

```python
# Sorting buses based on number of passengers

df.sort_values('Passengers', ascending=False)
```

```python
# Calculating estimated revenue using passengers and ticket fare

df['Revenue'] = df['Passengers'] * df['Ticket_Fare']

df.head()
```

```python
# Calculating total estimated revenue

df['Revenue'].sum()
```

```python
# Finding the route with the highest estimated revenue

df.loc[df['Revenue'].idxmax()]
```

```python
# Creating a bar chart for passengers on each route

plt.figure(figsize=(12,5))

plt.bar(df['Route'], df['Passengers'])

plt.xlabel('Route')
plt.ylabel('Number of Passengers')
plt.title('Passengers by KSRTC Route')

plt.show()
```

```python
# Creating a line chart for distance travelled by each route

plt.figure(figsize=(12,5))

plt.plot(df['Route'], df['Distance_km'], marker='o')

plt.xlabel('Route')
plt.ylabel('Distance (km)')
plt.title('Distance by KSRTC Route')

plt.show()
```

```python
# Creating a histogram to show passenger distribution

plt.hist(df['Passengers'], bins=5)

plt.xlabel('Number of Passengers')
plt.ylabel('Frequency')
plt.title('Passenger Distribution')

plt.show()
```

```python
# Creating a scatter plot to compare distance and passengers

plt.scatter(df['Distance_km'], df['Passengers'])

plt.xlabel('Distance (km)')
plt.ylabel('Passengers')
plt.title('Distance vs Passengers')

plt.show()
```

```python
# Finding the route with the highest fuel cost

df.loc[df['Fuel_Cost'].idxmax()]
```

```python
# Displaying important project results

print("Total Passengers:", df['Passengers'].sum())
print("Average Passengers:", df['Passengers'].mean())
print("Total Revenue:", df['Revenue'].sum())
print("Average Fuel Cost:", df['Fuel_Cost'].mean())
```

```python
# Importing Seaborn for data visualization

import seaborn as sns
```

```python
# Finding correlation between numerical columns

correlation = df[['Distance_km', 'Passengers', 'Ticket_Fare', 'Fuel_Cost', 'Revenue']].corr()

correlation
```

```python
# Creating a correlation heatmap using Seaborn

plt.figure(figsize=(8,6))

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title('Correlation Heatmap of KSRTC Dataset')

plt.show()
```

```python
# Creating a scatter plot using Seaborn

plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x='Distance_km', y='Passengers')

plt.xlabel('Distance (km)')
plt.ylabel('Passengers')
plt.title('Distance vs Passengers')

plt.show()
```
